```python
# importing pandas

import pandas as pd
```

```python
# importing CSV using read function

df = pd.read_csv("D:\Downloads\customer_shopping_behavior.csv")
df.head()
```

```python
# Studing the data set

df.info()
```

```python
# Checking duplicates

print(df[df.duplicated()])
print(df.duplicated().sum())
```

```python
# Checking NULL values

df.isnull().sum()
```

```python
# Populating NULLs with medion value for each category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.columns
```

```python
# Renaming columns into snake casing

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.head()
```

```python
# Creating age group column based on age

labels = ['Young-Adult','Adult','Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# Removing Redundant column

(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop(columns = 'promo_code_used')
```

```python
# Converting column to numerical value

df['frequency_of_purchases'].unique()
```

```python
freq_map = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(freq_map)
```
